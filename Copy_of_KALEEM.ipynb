```python
#/ EXPERIMENT 1
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1


def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid
        elif key < arr[mid]:
            high = mid - 1
        else:
            low = mid + 1

    return -1


print("1. Linear Search")
print("2. Binary Search")

choice = int(input("Enter your choice: "))

n = int(input("Enter size of array: "))
arr = []

for i in range(n):
    arr.append(int(input("Enter element: ")))

key = int(input("Enter key: "))

if choice == 1:
    result = linear_search(arr, key)
elif choice == 2:
    arr.sort()
    print("Sorted Array:", arr)
    result = binary_search(arr, key)
else:
    result = -1

if result == -1:
    print("Unsuccessful Search")
else:
    print(key, "found at position", result + 1)
```

```python
#/ EXPERIMENT 2
def bubble_sort(a):
    n = len(a)
    for i in range(n):
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
        print("Pass", i + 1, ":", a)


def selection_sort(a):
    n = len(a)
    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if a[j] < a[min_index]:
                min_index = j

        a[i], a[min_index] = a[min_index], a[i]
        print("Pass", i + 1, ":", a)


def insertion_sort(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1

        a[j + 1] = key
        print("Pass", i, ":", a)


def quick_sort(a):
    if len(a) <= 1:
        return a

    pivot = a[0]
    left = [x for x in a[1:] if x <= pivot]
    right = [x for x in a[1:] if x > pivot]

    return quick_sort(left) + [pivot] + quick_sort(right)


def merge_sort(a):
    if len(a) <= 1:
        return a

    mid = len(a) // 2

    left = merge_sort(a[:mid])
    right = merge_sort(a[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


while True:
    print("\n--- SORTING ---")
    print("1. Bubble Sort")
    print("2. Selection Sort")
    print("3. Insertion Sort")
    print("4. Quick Sort")
    print("5. Merge Sort")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 6:
        break

    n = int(input("Enter size of array: "))
    arr = []

    for i in range(n):
        arr.append(int(input("Enter element: ")))

    if choice == 1:
        bubble_sort(arr)
    elif choice == 2:
        selection_sort(arr)
    elif choice == 3:
        insertion_sort(arr)
    elif choice == 4:
        arr = quick_sort(arr)
    elif choice == 5:
        arr = merge_sort(arr)
    else:
        print("Invalid choice")
        continue

    print("Sorted Array:", arr)
```

```python
#/ EXPERIMENT 4
def infix_to_postfix(expression):
    stack = []
    output = []

    precedence = {
        '+': 1,
        '-': 1,
        '*': 2,
        '/': 2,
        '^': 3
    }

    for ch in expression:
        if ch.isalnum():
            output.append(ch)

        elif ch == '(':
            stack.append(ch)

        elif ch == ')':
            while stack and stack[-1] != '(':
                output.append(stack.pop())
            if not stack:
                print("Invalid Expression")
                return
            stack.pop()

        elif ch in precedence:
            while (stack and stack[-1] != '(' and
                   precedence.get(stack[-1], 0) >= precedence[ch]):
                output.append(stack.pop())

            stack.append(ch)

    while stack:
        if stack[-1] == '(':
            print("Invalid Expression")
            return
        output.append(stack.pop())

    print("Postfix Expression:", ''.join(output))


def evaluate_postfix(expression):
    stack = []

    for ch in expression:
        if ch.isdigit():
            stack.append(int(ch))

        elif ch in "+-*/^":
            if len(stack) < 2:
                print("Invalid Expression")
                return

            b = stack.pop()
            a = stack.pop()

            if ch == '+':
                stack.append(a + b)
            elif ch == '-':
                stack.append(a - b)
            elif ch == '*':
                stack.append(a * b)
            elif ch == '/':
                stack.append(a / b)
            elif ch == '^':
                stack.append(a ** b)

    if len(stack) == 1:
        print("Evaluation Result:", stack.pop())
    else:
        print("Invalid Expression")


while True:
    print("\n--- EXPRESSION ---")
    print("1. Infix to Postfix")
    print("2. Evaluate Postfix")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        exp = input("Enter infix expression: ")
        infix_to_postfix(exp)

    elif choice == 2:
        exp = input("Enter postfix expression: ")
        evaluate_postfix(exp)

    elif choice == 3:
        break

    else:
        print("Invalid choice")
```

```python
#/ EXPERIMENT 5
class CircularQueue:
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = -1
        self.rear = -1

    def enqueue(self, item):
        if (self.rear + 1) % self.size == self.front:
            print("Queue Overflow")
            return

        if self.front == -1:
            self.front = 0

        self.rear = (self.rear + 1) % self.size
        self.queue[self.rear] = item

        print("Item inserted successfully")

    def dequeue(self):
        if self.front == -1:
            print("Queue Underflow")
            return

        item = self.queue[self.front]
        print("Deleted item:", item)

        if self.front == self.rear:
            self.front = -1
            self.rear = -1
        else:
            self.front = (self.front + 1) % self.size

    def peek(self):
        if self.front == -1:
            print("Queue Underflow")
        else:
            print("Front item:", self.queue[self.front])

    def display(self):
        if self.front == -1:
            print("Queue Underflow")
            return

        print("Queue elements:", end=" ")

        i = self.front

        while True:
            print(self.queue[i], end=" ")

            if i == self.rear:
                break

            i = (i + 1) % self.size

        print()


size = int(input("Enter capacity of Circular Queue: "))

if size <= 0:
    print("Invalid capacity")
else:
    q = CircularQueue(size)

    while True:
        print("\n--- CIRCULAR QUEUE ---")
        print("1. Enqueue")
        print("2. Dequeue")
        print("3. Peek")
        print("4. Display")
        print("5. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            item = int(input("Enter item: "))
            q.enqueue(item)

        elif choice == 2:
            q.dequeue()

        elif choice == 3:
            q.peek()

        elif choice == 4:
            q.display()

        elif choice == 5:
            print("Exit")
            break

        else:
            print("Invalid choice")
```

```python
#/ EXPERIMENT 6
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        print("Node inserted")

    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
        else:
            temp = self.head

            while temp.next:
                temp = temp.next

            temp.next = new_node

        print("Node inserted")

    def insert_position(self, data, pos):
        if pos < 1:
            print("Invalid position")
            return

        if pos == 1:
            self.insert_beginning(data)
            return

        temp = self.head

        for i in range(pos - 2):
            if temp is None:
                print("Invalid position")
                return
            temp = temp.next

        if temp is None:
            print("Invalid position")
            return

        new_node = Node(data)
        new_node.next = temp.next
        temp.next = new_node

        print("Node inserted")

    def delete_beginning(self):
        if self.head is None:
            print("List is empty")
        else:
            self.head = self.head.next
            print("First node deleted")

    def delete_end(self):
        if self.head is None:
            print("List is empty")
            return

        if self.head.next is None:
            self.head = None
        else:
            temp = self.head

            while temp.next.next:
                temp = temp.next

            temp.next = None

        print("Last node deleted")

    def delete_value(self, value):
        if self.head is None:
            print("List is empty")
            return

        if self.head.data == value:
            self.head = self.head.next
            print("Element deleted")
            return

        temp = self.head

        while temp.next and temp.next.data != value:
            temp = temp.next

        if temp.next is None:
            print("Element not found")
        else:
            temp.next = temp.next.next
            print("Element deleted")

    def search(self, value):
        temp = self.head
        pos = 1

        while temp:
            if temp.data == value:
                print("Element found at position", pos)
                return

            temp = temp.next
            pos += 1

        print("Element not found")

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


sll = SinglyLinkedList()

while True:
    print("\n--- SINGLY LINKED LIST ---")
    print("1. Insert at Beginning")
    print("2. Insert at End")
    print("3. Insert at Specific Position")
    print("4. Delete from Beginning")
    print("5. Delete from End")
    print("6. Delete Specific Element")
    print("7. Search")
    print("8. Display")
    print("9. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        data = int(input("Enter data: "))
        sll.insert_beginning(data)

    elif choice == 2:
        data = int(input("Enter data: "))
        sll.insert_end(data)

    elif choice == 3:
        data = int(input("Enter data: "))
        pos = int(input("Enter position: "))
        sll.insert_position(data, pos)

    elif choice == 4:
        sll.delete_beginning()

    elif choice == 5:
        sll.delete_end()

    elif choice == 6:
        data = int(input("Enter element to delete: "))
        sll.delete_value(data)

    elif choice == 7:
        data = int(input("Enter element to search: "))
        sll.search(data)

    elif choice == 8:
        sll.display()

    elif choice == 9:
        print("Exit")
        break

    else:
        print("Invalid choice")
```

```python
#/ EXPERIMENT 7
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None

    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head

        if self.head:
            self.head.prev = new_node

        self.head = new_node
        print("Node inserted")

    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
        else:
            temp = self.head

            while temp.next:
                temp = temp.next

            temp.next = new_node
            new_node.prev = temp

        print("Node inserted")

    def insert_position(self, data, pos):
        if pos < 1:
            print("Invalid position")
            return

        if pos == 1:
            self.insert_beginning(data)
            return

        temp = self.head

        for i in range(pos - 2):
            if temp is None:
                print("Invalid position")
                return
            temp = temp.next

        if temp is None:
            print("Invalid position")
            return

        new_node = Node(data)
        new_node.next = temp.next
        new_node.prev = temp

        if temp.next:
            temp.next.prev = new_node

        temp.next = new_node
        print("Node inserted")

    def delete_beginning(self):
        if self.head is None:
            print("List is empty")
            return

        self.head = self.head.next

        if self.head:
            self.head.prev = None

        print("First node deleted")

    def delete_end(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head

        while temp.next:
            temp = temp.next

        if temp.prev:
            temp.prev.next = None
        else:
            self.head = None

        print("Last node deleted")

    def delete_value(self, value):
        temp = self.head

        while temp and temp.data != value:
            temp = temp.next

        if temp is None:
            print("Element not found")
            return

        if temp.prev:
            temp.prev.next = temp.next
        else:
            self.head = temp.next

        if temp.next:
            temp.next.prev = temp.prev

        print("Element deleted")

    def search(self, value):
        temp = self.head
        pos = 1

        while temp:
            if temp.data == value:
                print("Element found at position", pos)
                return

            temp = temp.next
            pos += 1

        print("Element not found")

    def display_forward(self):
        temp = self.head

        if temp is None:
            print("List is empty")
            return

        while temp:
            print(temp.data, end=" <-> ")
            temp = temp.next

        print("None")

    def display_backward(self):
        temp = self.head

        if temp is None:
            print("List is empty")
            return

        while temp.next:
            temp = temp.next

        while temp:
            print(temp.data, end=" <-> ")
            temp = temp.prev

        print("None")


dll = DoublyLinkedList()

while True:
    print("\n--- DOUBLY LINKED LIST ---")
    print("1. Insert at Beginning")
    print("2. Insert at End")
    print("3. Insert at Specific Position")
    print("4. Delete from Beginning")
    print("5. Delete from End")
    print("6. Delete Specific Element")
    print("7. Search")
    print("8. Display Forward")
    print("9. Display Backward")
    print("10. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        data = int(input("Enter data: "))
        dll.insert_beginning(data)

    elif choice == 2:
        data = int(input("Enter data: "))
        dll.insert_end(data)

    elif choice == 3:
        data = int(input("Enter data: "))
        pos = int(input("Enter position: "))
        dll.insert_position(data, pos)

    elif choice == 4:
        dll.delete_beginning()

    elif choice == 5:
        dll.delete_end()

    elif choice == 6:
        data = int(input("Enter element to delete: "))
        dll.delete_value(data)

    elif choice == 7:
        data = int(input("Enter element to search: "))
        dll.search(data)

    elif choice == 8:
        dll.display_forward()

    elif choice == 9:
        dll.display_backward()

    elif choice == 10:
        print("Exit")
        break

    else:
        print("Invalid choice")
```

```python
#/ EXPERIMENT 8
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    def insert_beginning(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
        else:
            temp = self.head

            while temp.next != self.head:
                temp = temp.next

            new_node.next = self.head
            temp.next = new_node
            self.head = new_node

        print("Node inserted")

    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
        else:
            temp = self.head

            while temp.next != self.head:
                temp = temp.next

            temp.next = new_node
            new_node.next = self.head

        print("Node inserted")

    def insert_position(self, data, pos):
        if pos < 1:
            print("Invalid position")
            return

        if pos == 1:
            self.insert_beginning(data)
            return

        if self.head is None:
            print("Invalid position")
            return

        temp = self.head

        for i in range(pos - 2):
            temp = temp.next

            if temp == self.head:
                print("Invalid position")
                return

        new_node = Node(data)
        new_node.next = temp.next
        temp.next = new_node

        print("Node inserted")

    def delete_beginning(self):
        if self.head is None:
            print("List is empty")
            return

        if self.head.next == self.head:
            self.head = None
        else:
            temp = self.head

            while temp.next != self.head:
                temp = temp.next

            self.head = self.head.next
            temp.next = self.head

        print("First node deleted")

    def delete_end(self):
        if self.head is None:
            print("List is empty")
            return

        if self.head.next == self.head:
            self.head = None
        else:
            temp = self.head

            while temp.next.next != self.head:
                temp = temp.next

            temp.next = self.head

        print("Last node deleted")

    def delete_position(self, pos):
        if self.head is None:
            print("List is empty")
            return

        if pos < 1:
            print("Invalid position")
            return

        if pos == 1:
            self.delete_beginning()
            return

        prev = self.head
        temp = self.head.next
        count = 2

        while temp != self.head and count < pos:
            prev = temp
            temp = temp.next
            count += 1

        if temp == self.head:
            print("Invalid position")
            return

        prev.next = temp.next
        print("Node deleted")

    def search(self, value):
        if self.head is None:
            print("Element not found")
            return

        temp = self.head
        pos = 1

        while True:
            if temp.data == value:
                print("Element found at position", pos)
                return

            temp = temp.next
            pos += 1

            if temp == self.head:
                break

        print("Element not found")

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head

        while True:
            print(temp.data, end=" -> ")
            temp = temp.next

            if temp == self.head:
                break

        print("(back to head)")


cll = CircularLinkedList()

while True:
    print("\n--- CIRCULAR LINKED LIST ---")
    print("1. Insert at Beginning")
    print("2. Insert at End")
    print("3. Insert at Specific Position")
    print("4. Delete from Beginning")
    print("5. Delete from End")
    print("6. Delete at Specific Position")
    print("7. Search")
    print("8. Display")
    print("9. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        data = int(input("Enter data: "))
        cll.insert_beginning(data)

    elif choice == 2:
        data = int(input("Enter data: "))
        cll.insert_end(data)

    elif choice == 3:
        data = int(input("Enter data: "))
        pos = int(input("Enter position: "))
        cll.insert_position(data, pos)

    elif choice == 4:
        cll.delete_beginning()

    elif choice == 5:
        cll.delete_end()

    elif choice == 6:
        pos = int(input("Enter position to delete: "))
        cll.delete_position(pos)

    elif choice == 7:
        data = int(input("Enter element to search: "))
        cll.search(data)

    elif choice == 8:
        cll.display()

    elif choice == 9:
        print("Exit")
        break

    else:
        print("Invalid choice")
```

```python
#/ EXPERIMENT 9

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:
    def __init__(self):
        self.top = None

    def isEmpty(self):
        return self.top is None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        print("Item pushed successfully")

    def pop(self):
        if self.isEmpty():
            print("Stack Underflow")
        else:
            item = self.top.data
            self.top = self.top.next
            print("Popped item is", item)

    def peek(self):
        if self.isEmpty():
            print("Stack Underflow")
        else:
            print("Top item is", self.top.data)

    def display(self):
        if self.isEmpty():
            print("Stack Underflow")
            return

        temp = self.top
        print("Stack elements:")

        while temp is not None:
            print(temp.data)
            temp = temp.next


stack = Stack()

while True:
    print("\n--- STACK USING LINKED LIST ---")
    print("1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Display")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        item = int(input("Enter item to push: "))
        stack.push(item)

    elif choice == '2':
        stack.pop()

    elif choice == '3':
        stack.peek()

    elif choice == '4':
        stack.display()

    elif choice == '5':
        print("Exit Program")
        break

    else:
        print("Invalid choice")
```

```python
# EXPERIMENT 10

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return Node(data)

        if data < root.data:
            root.left = self.insert(root.left, data)
        elif data > root.data:
            root.right = self.insert(root.right, data)
        else:
            print("Duplicate elements are not allowed")

        return root

    def search(self, root, key):
        if root is None:
            return False

        if root.data == key:
            return True
        elif key < root.data:
            return self.search(root.left, key)
        else:
            return self.search(root.right, key)

    def find_min(self, root):
        while root.left is not None:
            root = root.left
        return root

    def delete(self, root, key):
        if root is None:
            return None

        if key < root.data:
            root.left = self.delete(root.left, key)

        elif key > root.data:
            root.right = self.delete(root.right, key)

        else:
            # Case 1: No left child
            if root.left is None:
                return root.right

            # Case 2: No right child
            if root.right is None:
                return root.left

            # Case 3: Two children
            successor = self.find_min(root.right)
            root.data = successor.data
            root.right = self.delete(root.right, successor.data)

        return root

    def inorder(self, root):
        if root:
            self.inorder(root.left)
            print(root.data, end=" ")
            self.inorder(root.right)


bst = BinarySearchTree()

while True:
    print("\n--- BINARY SEARCH TREE ---")
    print("1. Insertion")
    print("2. Deletion")
    print("3. Search")
    print("4. Display (Inorder)")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        value = int(input("Enter value to insert: "))
        if bst.search(bst.root, value):
            print("Duplicate elements are not allowed")
        else:
            bst.root = bst.insert(bst.root, value)
            print("Element inserted successfully")

    elif choice == '2':
        value = int(input("Enter value to delete: "))
        if bst.search(bst.root, value):
            bst.root = bst.delete(bst.root, value)
            print("Element deleted successfully")
        else:
            print("Element not found")

    elif choice == '3':
        value = int(input("Enter value to search: "))
        if bst.search(bst.root, value):
            print("Element found")
        else:
            print("Element not found")

    elif choice == '4':
        print("Inorder Traversal:")
        bst.inorder(bst.root)
        print()

    elif choice == '5':
        print("Exit Program")
        break

    else:
        print("Invalid choice")
```
